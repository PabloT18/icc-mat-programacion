## Referencia rápida de leyes

| Ley | Forma `and` | Forma `or` |
|-----|-------------|------------|
| **Identidad** | `A and True` ≡ `A` | `A or False` ≡ `A` |
| **Dominación** | `A and False` ≡ `False` | `A or True` ≡ `True` |
| **Idempotencia** | `A and A` ≡ `A` | `A or A` ≡ `A` |
| **Complemento** | `A and not A` ≡ `False` | `A or not A` ≡ `True` |
| **Doble Negación** | `not (not A)` ≡ `A` | — |
| **Conmutativa** | `A and B` ≡ `B and A` | `A or B` ≡ `B or A` |
| **Asociativa** | `(A and B) and C` ≡ `A and B and C` | `(A or B) or C` ≡ `A or B or C` |
| **Distributiva** | `A and (B or C)` ≡ `(A and B) or (A and C)` | `A or (B and C)` ≡ `(A or B) and (A or C)` |
| **De Morgan** | `not (A and B)` ≡ `not A or not B` | `not (A or B)` ≡ `not A and not B` |
| **Absorción** | `A and (A or B)` ≡ `A` | `A or (A and B)` ≡ `A` |
