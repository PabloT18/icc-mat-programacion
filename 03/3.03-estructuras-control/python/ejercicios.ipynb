## Ejercicio 01. Signo

Clasificar -4 como positivo, negativo o cero.

### Solución


```python
n=-4
if n>0: print("Positivo")
elif n<0: print("Negativo")
else: print("Cero")
```

## Ejercicio 03. Mayor de dos

Comparar 7 y 7. Mostrar el mayor o “Iguales” cuando coincidan.

### Solución


```python
a,b=7,7
if a>b: print(a)
elif b>a: print(b)
else: print("Iguales")
```

## Ejercicio 07. Clasificación de edad

Clasificar 31 en menor de 18, de 18 a 30, de 31 a 50 o mayor de 50. Rechazar negativos.

### Solución


```python
edad=31
if edad<0: print("Inválida")
elif edad<18: print("Menor de 18")
elif edad<=30: print("18 a 30")
elif edad<=50: print("31 a 50")
else: print("Mayor de 50")
```

## Ejercicio 08. Año bisiesto

Determinar si 1900 es bisiesto: divisible por 400 o divisible por 4 y no por 100.

### Solución


```python
anio=1900
bisiesto=anio%400==0 or (anio%4==0 and anio%100!=0)
print("Bisiesto" if bisiesto else "No bisiesto")
```

## Ejercicio 09. Triángulo válido

Recibir lados 3, 4 y 5. Rechazar lados no positivos o que incumplan desigualdad triangular. Clasificar equilátero, isósceles o escaleno.

### Solución


```python
a,b,c=3,4,5
if min(a,b,c)<=0 or a+b<=c or a+c<=b or b+c<=a: print("Inválido")
elif a==b==c: print("Equilátero")
elif a==b or a==c or b==c: print("Isósceles")
else: print("Escaleno")
```

## Ejercicio 10. Entrada y rango

Convertir el texto “abc” a nota entera entre 0 y 10. Distinguir formato inválido, rango inválido y valor aceptado.

### Solución


```python
texto="abc"
try:
    nota=int(texto)
    if 0<=nota<=10: print("Aceptada")
    else: print("Fuera de rango")
except ValueError:
    print("Formato inválido")
```
