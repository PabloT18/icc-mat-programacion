## Ejercicio 01. Escritura UTF-8

Escribir “Cuenca” en un archivo temporal UTF-8 y leerlo de nuevo.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("Cuenca",encoding="utf-8")
    print(ruta.read_text(encoding="utf-8"))
```

## Ejercicio 02. Dos líneas

Guardar y leer dos líneas: sol y luna. Conservar un salto final.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("sol\nluna\n",encoding="utf-8")
    print(ruta.read_text(encoding="utf-8"),end="")
```

## Ejercicio 03. Anexado

Crear una línea A y anexar B sin eliminar A.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("A\n",encoding="utf-8")
    with ruta.open("a",encoding="utf-8") as archivo: archivo.write("B\n")
    print(ruta.read_text(encoding="utf-8"),end="")
```

## Ejercicio 04. Contar líneas no vacías

Contar líneas con texto en “a”, vacía y “b”. Ignorar espacios externos.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("a\n  \nb\n",encoding="utf-8")
    contador=0
    with ruta.open(encoding="utf-8") as archivo:
        for linea in archivo:
            if linea.strip(): contador+=1
    print(contador)
```

## Ejercicio 05. Suma de archivo

Leer enteros 4, 8 y 3, uno por línea, y mostrar suma.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("4\n8\n3\n",encoding="utf-8")
    total=0
    with ruta.open(encoding="utf-8") as archivo:
        for linea in archivo: total+=int(linea)
    print(total)
```

## Ejercicio 06. Archivo inexistente

Intentar leer ausente.txt dentro de una carpeta temporal y producir mensaje controlado.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    try:
        (base/"ausente.txt").read_text(encoding="utf-8")
    except FileNotFoundError:
        print("Archivo no encontrado")
```

## Ejercicio 07. Errores por línea

Leer 4, x y 6. Sumar válidos y registrar número de línea inválida.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("4\nx\n6\n",encoding="utf-8")
    total=0
    with ruta.open(encoding="utf-8") as archivo:
        for numero,linea in enumerate(archivo,1):
            try: total+=int(linea)
            except ValueError: print(f"Línea {numero}: inválida")
    print(f"Total: {total}")
```

## Ejercicio 08. Filtrar a otro archivo

Leer 2, 7, 4 y escribir solo valores mayores que 3 en un archivo nuevo. Conservar entrada.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    origen=base/"entrada.txt"; destino=base/"salida.txt"
    texto="2\n7\n4\n"
    origen.write_text(texto,encoding="utf-8")
    with origen.open(encoding="utf-8") as entrada, destino.open("w",encoding="utf-8") as salida:
        for linea in entrada:
            if int(linea)>3: salida.write(linea)
    assert origen.read_text(encoding="utf-8")==texto
    print(destino.read_text(encoding="utf-8"),end="")
```

## Ejercicio 09. Promedio sin datos

Leer archivo vacío y mostrar “Sin datos” sin dividir por cero.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("",encoding="utf-8")
    datos=[]
    with ruta.open(encoding="utf-8") as archivo:
        for linea in archivo:
            if linea.strip(): datos.append(float(linea))
    if datos: print(sum(datos)/len(datos))
    else: print("Sin datos")
```

## Ejercicio 10. Registros delimitados

Leer A;3 y B;5. Exigir exactamente dos campos, código no vacío y cantidad entera no negativa. Mostrar suma y conservar número de línea en errores. El formato no permite ; dentro de campos.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    ruta=base/"dato.txt"
    ruta.write_text("A;3\nB;5\n",encoding="utf-8")
    total=0
    with ruta.open(encoding="utf-8") as archivo:
        for numero,linea in enumerate(archivo,1):
            campos=linea.strip().split(";")
            if len(campos)!=2 or not campos[0]: raise ValueError(f"Línea {numero}: formato")
            cantidad=int(campos[1])
            if cantidad<0: raise ValueError(f"Línea {numero}: cantidad")
            total+=cantidad
    print(total)
```
