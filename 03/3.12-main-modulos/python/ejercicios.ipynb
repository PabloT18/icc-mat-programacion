## Ejercicio 06. Error en la frontera

Definir dividir que lance error para divisor cero. Gestionarlo en main con datos 4 y 0.

### Solución


```python
def dividir(a,b):
    if b==0: raise ValueError("Divisor cero")
    return a/b
def main():
    try: print(dividir(4,0))
    except ValueError as error: print(error)
if __name__=="__main__": main()
```

## Ejercicio 07. Módulo utilitario real

Crear temporalmente operaciones.py con suma(a,b), importarlo sin salida y utilizar suma(3,4). En Java crear Operaciones.java y compilar ambos fuentes.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import importlib.util
    ruta=base/"operaciones.py"
    ruta.write_text("def suma(a,b):\n    return a+b\n",encoding="utf-8")
    spec=importlib.util.spec_from_file_location("operaciones_ejercicio",ruta)
    modulo=importlib.util.module_from_spec(spec)
    spec.loader.exec_module(modulo)
    print(modulo.suma(3,4))
```

## Ejercicio 08. Importación sin interacción

Crear un módulo con main protegido. Importarlo y comprobar que no produjo salida. Mostrar “Importación limpia” al terminar.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import importlib.util,io,contextlib
    ruta=base/"entrada.py"
    ruta.write_text("def main():\n    print(123)\nif __name__ == '__main__':\n    main()\n",encoding="utf-8")
    spec=importlib.util.spec_from_file_location("entrada_ejercicio",ruta)
    modulo=importlib.util.module_from_spec(spec)
    captura=io.StringIO()
    with contextlib.redirect_stdout(captura): spec.loader.exec_module(modulo)
    assert captura.getvalue()==""
    print("Importación limpia")
```

## Ejercicio 09. Integración de funciones

Filtrar pares, elevarlos al cuadrado y mostrarlos desde main para [1,2,3,4]. Separar tres responsabilidades.

### Solución


```python
def filtrar(datos): return [n for n in datos if n%2==0]
def cuadrados(datos): return [n*n for n in datos]
def mostrar(datos): print(*datos)
def main(): mostrar(cuadrados(filtrar([1,2,3,4])))
if __name__=="__main__": main()
```

## Ejercicio 10. Ejecución en proceso nuevo

Crear temporalmente util.py y main.py. Ejecutar main.py en un proceso nuevo y obtener “Total: 6” para [1,2,3]. En Java compilar utilidades y entrada juntas.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import subprocess,sys
    (base/"util.py").write_text("def total(datos):\n    return sum(datos)\n",encoding="utf-8")
    (base/"main.py").write_text("from util import total\ndef main():\n    print(f'Total: {total([1,2,3])}')\nif __name__ == '__main__':\n    main()\n",encoding="utf-8")
    resultado=subprocess.run([sys.executable,"main.py"],cwd=base,text=True,capture_output=True,check=True)
    print(resultado.stdout,end="")
```
