## Ejercicio 04. Máximo de tres

Definir mayor(a,b,c) sin usar max. Probar 8, 3 y 8.

### Solución


```python
def mayor(a,b,c):
    resultado=a
    if b>resultado: resultado=b
    if c>resultado: resultado=c
    return resultado
print(mayor(8,3,8))
```

## Ejercicio 05. Suma de rango

Definir sumar_rango(inicio,fin) inclusivo. Rechazar inicio mayor que fin. Probar 3 a 6.

### Solución


```python
def sumar_rango(inicio,fin):
    if inicio>fin: raise ValueError("Intervalo inválido")
    total=0
    for n in range(inicio,fin+1): total+=n
    return total
print(sumar_rango(3,6))
```

## Ejercicio 06. Factorial validado

Definir factorial(n) para enteros no negativos. Probar cero y cinco.

### Solución


```python
def factorial(n):
    if n<0: raise ValueError("Negativo")
    resultado=1
    for i in range(1,n+1): resultado*=i
    return resultado
print(factorial(0),factorial(5))
```

## Ejercicio 07. Contar divisores

Definir contar_divisores(n) para n>0. Probar 12.

### Solución


```python
def contar_divisores(n):
    if n<=0: raise ValueError("Se requiere positivo")
    cantidad=0
    for d in range(1,n+1):
        if n%d==0: cantidad+=1
    return cantidad
print(contar_divisores(12))
```

## Ejercicio 08. Número perfecto

Definir es_perfecto(n) utilizando suma de divisores propios. Probar 6 y 8.

### Solución


```python
def es_perfecto(n):
    if n<2: return False
    suma=0
    for d in range(1,n):
        if n%d==0: suma+=d
    return suma==n
print(es_perfecto(6),es_perfecto(8))
```

## Ejercicio 09. Máximo común divisor

Definir mcd(a,b) mediante Euclides. Aceptar enteros, usar valores absolutos y definir mcd(0,0)=0. Probar 48 y 18.

### Solución


```python
def mcd(a,b):
    a,b=abs(a),abs(b)
    while b!=0: a,b=b,a%b
    return a
print(mcd(48,18))
```

## Ejercicio 10. Composición

Definir costo(cantidad,precio) y aplicar_descuento(total,porcentaje). Validar dominios y componerlas para 4 artículos a 12 con 25% de descuento.

### Solución


```python
def costo(cantidad,precio):
    if cantidad<0 or precio<0: raise ValueError("Negativo")
    return cantidad*precio
def aplicar_descuento(total,porcentaje):
    if total<0 or not 0<=porcentaje<=100: raise ValueError("Dominio inválido")
    return total*(1-porcentaje/100)
print(f"{aplicar_descuento(costo(4,12),25):.2f}")
```
