## Ejercicio 01. Saludo

Leer un nombre y mostrar “Hola, Ana” para la entrada Ana. Conservar los espacios interiores del nombre.

### Solución


```python
from io import StringIO
entrada = StringIO("Ana\n")
nombre = entrada.readline().rstrip("\n")
print(f"Hola, {nombre}")
```

## Ejercicio 02. Suma de enteros

Leer dos enteros, uno por línea. Para 12 y 8 mostrar su suma.

### Solución


```python
from io import StringIO
entrada = StringIO("12\n8\n")
a = int(entrada.readline())
b = int(entrada.readline())
print(a+b)
```

## Ejercicio 03. Nombre y edad

Leer nombre y edad. Para Luis y 19 mostrar “Luis: 19 años”. Convertir edad a entero.

### Solución


```python
from io import StringIO
entrada = StringIO("Luis\n19\n")
nombre = entrada.readline().strip()
edad = int(entrada.readline())
print(f"{nombre}: {edad} años")
```

## Ejercicio 04. Área de triángulo

Leer base y altura positivas. Para 10.5 y 4 mostrar área con dos decimales.

### Solución


```python
from io import StringIO
entrada = StringIO("10.5\n4\n")
base = float(entrada.readline())
altura = float(entrada.readline())
print(f"{base*altura/2:.2f}")
```

## Ejercicio 05. Temperatura

Leer grados Celsius y convertir a Fahrenheit. Para 25 mostrar 77.0.

### Solución


```python
from io import StringIO
entrada = StringIO("25\n")
celsius = float(entrada.readline())
print(f"{celsius*9/5+32:.1f}")
```

## Ejercicio 06. Identificador

Leer código 007 y cantidad 3. Conservar los ceros del código y mostrar “007: 3”.

## Ejercicio 07. Cociente decimal

Leer dividendo 7 y divisor 2, con divisor distinto de cero como precondición. Mostrar el cociente con tres decimales.

### Solución


```python
from io import StringIO
entrada = StringIO("7\n2\n")
a = float(entrada.readline())
b = float(entrada.readline())
print(f"{a/b:.3f}")
```

## Ejercicio 08. Promedio de mediciones

Leer tres decimales 8.5, 7.8 y 9.2. Mostrar promedio con dos decimales.

### Solución


```python
from io import StringIO
entrada = StringIO("8.5\n7.8\n9.2\n")
a = float(entrada.readline())
b = float(entrada.readline())
c = float(entrada.readline())
print(f"{(a+b+c)/3:.2f}")
```

## Ejercicio 09. Total de servicio

Leer horas 3.5, tarifa 8 y recargo porcentual 10. Calcular base, recargo y total, cada uno con dos decimales.

### Solución


```python
from io import StringIO
entrada = StringIO("3.5\n8\n10\n")
horas = float(entrada.readline())
tarifa = float(entrada.readline())
porcentaje = float(entrada.readline())
base = horas*tarifa
recargo = base*porcentaje/100
print(f"{base:.2f} {recargo:.2f} {base+recargo:.2f}")
```

## Ejercicio 10. Factura de dos productos

Leer nombre, cantidad y precio de dos productos. Para A, 2, 3.5 y B, 3, 2 mostrar un subtotal por línea y luego el total. No usar ciclos todavía.

### Solución


```python
from io import StringIO
entrada = StringIO("A\n2\n3.5\nB\n3\n2\n")
a=entrada.readline().strip()
qa=int(entrada.readline()); pa=float(entrada.readline())
b=entrada.readline().strip()
qb=int(entrada.readline()); pb=float(entrada.readline())
sa=qa*pa; sb=qb*pb
print(f"{a}: {sa:.2f}")
print(f"{b}: {sb:.2f}")
print(f"Total: {sa+sb:.2f}")
```
