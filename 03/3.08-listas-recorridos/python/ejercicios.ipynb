## Ejercicio 01. Suma explícita

Sumar [3,5,2] con un ciclo, sin sum.

### Solución


```python
total=0
for n in [3,5,2]: total+=n
print(total)
```

## Ejercicio 02. Cantidad de negativos

Contar negativos de [-2,0,4,-1].

### Solución


```python
cantidad=0
for n in [-2,0,4,-1]:
    if n<0: cantidad+=1
print(cantidad)
```

## Ejercicio 03. Máximo sin cero artificial

Calcular máximo de [-8,-2,-5]. Rechazar lista vacía.

### Solución


```python
datos=[-8,-2,-5]
if not datos: raise ValueError("Sin datos")
mayor=datos[0]
for n in datos[1:]:
    if n>mayor: mayor=n
print(mayor)
```

## Ejercicio 04. Primera posición

Retornar primera posición de 7 en [4,7,7], o -1 si no existe.

### Solución


```python
def buscar(datos,objetivo):
    for i,n in enumerate(datos):
        if n==objetivo: return i
    return -1
print(buscar([4,7,7],7))
```

## Ejercicio 05. Todas las posiciones

Obtener índices de 4 en [4,1,4,4].

### Solución


```python
resultado=[]
for i,n in enumerate([4,1,4,4]):
    if n==4: resultado.append(i)
print(*resultado)
```

## Ejercicio 06. Filtrado de pares

Seleccionar pares de [3,2,4,2,5] conservando orden y duplicados.

### Solución


```python
resultado=[]
for n in [3,2,4,2,5]:
    if n%2==0: resultado.append(n)
print(*resultado)
```

## Ejercicio 07. Transformar por signo

Transformar negativos a cero y duplicar los demás en [-2,0,3]. Conservar la entrada.

### Solución


```python
datos=[-2,0,3]
resultado=[]
for n in datos: resultado.append(0 if n<0 else n*2)
print(*resultado)
```

## Ejercicio 08. Eliminar duplicados estables

Eliminar repetidos de [3,1,3,2,1], conservando primera aparición. No usar conjuntos.

### Solución


```python
resultado=[]
for n in [3,1,3,2,1]:
    if n not in resultado: resultado.append(n)
print(*resultado)
```

## Ejercicio 09. Acumulados parciales

Producir [2,7,8] desde [2,5,1], donde cada posición contiene la suma hasta ese lugar.

### Solución


```python
total=0
resultado=[]
for n in [2,5,1]:
    total+=n
    resultado.append(total)
print(*resultado)
```

## Ejercicio 10. Mezcla ordenada

Mezclar [1,4,7] y [2,4,8] en orden, conservando duplicados, sin ordenar el resultado después.

### Solución


```python
def mezclar(a,b):
    i=j=0; resultado=[]
    while i<len(a) and j<len(b):
        if a[i]<=b[j]: resultado.append(a[i]); i+=1
        else: resultado.append(b[j]); j+=1
    resultado.extend(a[i:]); resultado.extend(b[j:])
    return resultado
print(*mezclar([1,4,7],[2,4,8]))
```
