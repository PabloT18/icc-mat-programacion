## Ejercicio 05. Comprobación sintáctica

Compilar sintácticamente un script que muestra 2+3, sin ejecutarlo. Guardar cualquier compilado en carpeta temporal.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import py_compile
    ruta=base/"demo.py"
    ruta.write_text("print(2+3)\n",encoding="utf-8")
    py_compile.compile(str(ruta),cfile=str(base/"demo.pyc"),doraise=True)
    print("Sintaxis válida")
```

## Ejercicio 06. Proceso independiente

Ejecutar un programa temporal que muestre 2+3. Capturar y mostrar su salida.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import subprocess,sys
    ruta=base/"demo.py"
    ruta.write_text("print(2+3)\n",encoding="utf-8")
    r=subprocess.run([sys.executable,str(ruta)],capture_output=True,text=True,check=True)
    print(r.stdout,end="")
```

## Ejercicio 07. Entrada automatizada

Ejecutar un programa nuevo que lea un entero y muestre su doble. Enviar 9 por entrada estándar y capturar la salida.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import subprocess,sys
    ruta=base/"demo.py"
    ruta.write_text("print(int(input())*2)\n",encoding="utf-8")
    r=subprocess.run([sys.executable,str(ruta)],input="9\n",capture_output=True,text=True,check=True)
    print(r.stdout,end="")
```

## Ejercicio 08. Código de error

Ejecutar un programa que termine con código 2. Detectar el código sin presentar la ejecución como exitosa.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import subprocess,sys
    ruta=base/"demo.py"
    ruta.write_text("raise SystemExit(2)\n",encoding="utf-8")
    r=subprocess.run([sys.executable,str(ruta)],capture_output=True,text=True)
    print(f"Código: {r.returncode}")
```

## Ejercicio 09. Comparación exacta

Ejecutar un programa que muestre “Total: 12”. Comparar stdout con “Total: 12” seguido de salto de línea y mostrar “Coincide”.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    import subprocess,sys
    ruta=base/"demo.py"
    ruta.write_text("print('Total: 12')\n",encoding="utf-8")
    r=subprocess.run([sys.executable,str(ruta)],capture_output=True,text=True,check=True)
    assert r.stdout=="Total: 12\n"
    print("Coincide")
```

## Ejercicio 10. Entrega limpia

Simular una carpeta con main.py, README.md, .venv y __pycache__. Obtener solo los archivos entregables y mostrar sus nombres ordenados. No borrar carpetas reales.

### Solución


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as carpeta:
    base = Path(carpeta)
    (base/"main.py").write_text("print(1)\n",encoding="utf-8")
    (base/"README.md").write_text("Ejecutar main.py",encoding="utf-8")
    (base/".venv").mkdir(); (base/"__pycache__").mkdir()
    archivos=sorted(p.name for p in base.iterdir() if p.is_file())
    print(" ".join(archivos))
```
