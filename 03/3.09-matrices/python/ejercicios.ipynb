## Ejercicio 03. Suma total

Sumar todas las celdas de [[1,2],[3,4]] con ciclos anidados.

### Solución


```python
total=0
for fila in [[1,2],[3,4]]:
    for n in fila: total+=n
print(total)
```

## Ejercicio 04. Sumas por fila

Producir una suma por fila de [[2,3],[4,7]].

### Solución


```python
resultado=[]
for fila in [[2,3],[4,7]]:
    suma=0
    for n in fila: suma+=n
    resultado.append(suma)
print(*resultado)
```

## Ejercicio 05. Sumas por columna

Sumar columnas de [[1,2,3],[4,5,6]], con rectangularidad como precondición.

### Solución


```python
m=[[1,2,3],[4,5,6]]
resultado=[0]*len(m[0])
for fila in m:
    for j,n in enumerate(fila): resultado[j]+=n
print(*resultado)
```

## Ejercicio 06. Diagonal principal

Sumar diagonal de [[1,2,3],[4,5,6],[7,8,9]]. Exigir matriz cuadrada.

### Solución


```python
m=[[1,2,3],[4,5,6],[7,8,9]]
if any(len(fila)!=len(m) for fila in m): raise ValueError("No cuadrada")
suma=0
for i in range(len(m)): suma+=m[i][i]
print(suma)
```

## Ejercicio 07. Transposición rectangular

Transponer [[1,2,3],[4,5,6]] a tres filas de dos elementos.

### Solución


```python
m=[[1,2,3],[4,5,6]]
resultado=[]
for j in range(len(m[0])):
    fila=[]
    for i in range(len(m)): fila.append(m[i][j])
    resultado.append(fila)
for fila in resultado: print(*fila)
```

## Ejercicio 08. Copia con umbral

Reemplazar por cero valores menores que 5 de [[2,8],[5,1]], devolviendo nueva matriz. Comprobar original intacto.

### Solución


```python
m=[[2,8],[5,1]]
resultado=[]
for fila in m:
    nueva=[]
    for n in fila: nueva.append(0 if n<5 else n)
    resultado.append(nueva)
assert m==[[2,8],[5,1]]
for fila in resultado: print(*fila)
```

## Ejercicio 09. Transformar palabras

Convertir a mayúsculas palabras con longitud al menos cuatro y conservar las otras en [["sol","luna"],["mar","cielo"]]. Crear filas independientes.

### Solución


```python
m=[["sol","luna"],["mar","cielo"]]
resultado=[]
for fila in m:
    nueva=[]
    for palabra in fila: nueva.append(palabra.upper() if len(palabra)>=4 else palabra)
    resultado.append(nueva)
for fila in resultado: print(" | ".join(fila))
```

## Ejercicio 10. Mayoría por fila

Para [[2,3,4],[1,3,2],[2,1]] mostrar cantidad de pares y clasificación “Mayoría par”, “Mayoría impar” o “Empate”. Admitir filas de distinto tamaño para este ejercicio.

### Solución


```python
for fila in [[2,3,4],[1,3,2],[2,1]]:
    pares=0
    for n in fila:
        if n%2==0: pares+=1
    impares=len(fila)-pares
    if pares>impares: etiqueta="Mayoría par"
    elif impares>pares: etiqueta="Mayoría impar"
    else: etiqueta="Empate"
    print(pares,etiqueta)
```
